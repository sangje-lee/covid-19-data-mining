Component required for the text mining

```python
# Component for NLTK library

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

# For tokenizer

from nltk.tokenize import RegexpTokenizer

# Finding frequency distinct in the text

from nltk.probability import FreqDist
```

Combine first 10000 title rows into a one big sentence

```python
s = ' '.join(df['title'].head(10000)) # (y_train['title'].head(10000))
print(s)
```

Download the required package for nltk Data Preprocessing

```python
# import nltk
# from nltk.corpus import stopwords
# from nltk.tokenize import word_tokenize
# from nltk.stem import PorterStemmer

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

Date preprocessing for first 10,000 rows for title without removal of punctuations in NLTK

```python
# Without remove punctuations in NLTK

# Text/Word contain 10,000 title
text = s

# Tokenization (Original)
tokens_original = word_tokenize(text)
print(tokens_original)

# Removing stopwords and stemming
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
tokens_cleaned_stopword = [stemmer.stem(word.lower()) for word in tokens_original if word.lower() not in stop_words]
print(tokens_cleaned_stopword)
```

```python
# from nltk.tokenize import RegexpTokenizer

# tokenizer = RegexpTokenizer(r'\w+')
# tokenizer.tokenize('Eighty-seven miles to go, yet.  Onward!')
```

Date preprocessing for first 10,000 rows for title "WITH" removal of punctuations in NLTK

```python
# Remove punctuations in NLTK

# Text/Word contain 10,000 title
text = s

# Tokenization with removing punctuations
tokenizer = RegexpTokenizer(r'\w+')
tokens2 = tokenizer.tokenize(text)
# print(tokens2)

# Removing stopwords and stemming plus removing punctuations
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
tokens_cleaned_all = [stemmer.stem(word.lower()) for word in tokens2 if word.lower() not in stop_words]
print(tokens_cleaned_all)
```

Finding frequency distinct in the text mining

```python
# fdist with punctual and no stopwords

fdist_original = FreqDist(tokens_original)
fdist_original
```

```python
# fdist with stopwords

fdist_cleaned_stopword = FreqDist(tokens_cleaned_stopword)
fdist_cleaned_stopword
```

```python
# fdist with removal of both punctual and stopwords

fdist_cleaned_all = FreqDist(tokens_cleaned_all)
fdist_cleaned_all
```

Converts a Python dictionary into a pandas DataFrame with "Frequency" column

```python
# Categorized into frequency columns wihtout stopword or removal of punctuation

df_fdist_original = pd.DataFrame.from_dict(fdist_original, orient='index', columns=['Frequency']) # Converts the dictionary 'fdist' into a pandas DataFrame.
print(df_fdist_original.head(10))                                   # Prints first ten rows of the DataFrame showing each item and its associated frequency.
```

```python
# Categorized into frequency columns based on cleaned the stopword

df_fdist_cleaned_stopword = pd.DataFrame.from_dict(fdist_cleaned_stopword, orient='index', columns=['Frequency'])
print(df_fdist_cleaned_stopword.head(10))
```

```python
# Categorized into frequency columns based on cleaned the stopword and removal of punctutions

df_fdist_cleaned_all = pd.DataFrame.from_dict(fdist_cleaned_all, orient='index', columns=['Frequency'])
print(df_fdist_cleaned_all.head(10))
```

Sort out from largest number to smallest based on the "Frequency" column

```python
# Sort out based on the frequency without any clean stopword or the punctuation

sorted_df_fdist_original = df_fdist_original.sort_values(by='Frequency', ascending=False) # Sort the dataset in decending order.

# print(sorted_df_fdist_original.head(100))                                                 # Prints first hundred rows of the DataFrame
```

```python
# Sort out based on the frequency based on cleaned the stopword

sorted_df_fdist_cleaned_stopword = df_fdist_cleaned_stopword.sort_values(by='Frequency', ascending=False) # Sort the dataset in decending order.

# print(sorted_df_fdist_cleaned_stopword.head(100))                                                         # Prints first hundred rows of the DataFrame
```

```python
# Sort out based on the frequency based on cleaned the stopword and removal of punctutions

sorted_df_fdist_cleaned_all  = df_fdist_cleaned_all.sort_values(by='Frequency', ascending=False) # Sort the dataset in decending order.

# print(sorted_df_fdist_cleaned_all.head(100))                                                     # Prints first hundred rows of the DataFrame
```
