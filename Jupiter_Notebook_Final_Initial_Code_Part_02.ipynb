Sklearn package including the tree and the train and test set split

```python
# from sklearn import tree
from sklearn.model_selection import train_test_split
```

Split the dataset into two pieces — a training set and a testing set.

```python
features = ['cord_uid','publish_time'] # X for 'cord_uid' and 'published date'
X = df.loc[:, features] 
y = df.loc[:, ['title', 'year']] # y for 'Title of the article' and 'published year of the article'
```

Consist of random sampling without replacement about 70 percent of the rows into your training set. The remaining 30 percent is put into your test set. 

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0, train_size = .70)
```
