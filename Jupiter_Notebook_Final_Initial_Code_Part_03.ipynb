Bagging/Emsemble Modelling

Train a machine learning model

```python
# from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train3 = scaler.fit_transform(X_train2)
y_train3 = y_train2.copy()
X_test3 = scaler.transform(X_test2)
y_test3 = y_test2.copy()
```

```python
# from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', DecisionTreeClassifier(random_state=42))
])

pipeline.fit(X_train3, y_train3)
```

Model evaluation

```python
# Make prediction on the testing data
y_pred_pipeline = pipeline.predict(X_test3)

# Classification Report
print(classification_report(y_pred_pipeline, y_test3))
```

```python
accuracy_pipeline = accuracy_score(y_test3, y_pred_pipeline)
precision_pipeline = precision_score(y_test3, y_pred_pipeline)
recall_pipeline = recall_score(y_test3, y_pred_pipeline)

print("Accuracy:", accuracy_pipeline)
print("Precision:", precision_pipeline)
print("Recall:", recall_pipeline)
```

```python
# Evaluate the classifier using cross-validation
# X_train3 should be X and y_train3 should be y.
cv_scores = cross_val_score(pipeline, X_train3, y_train3, cv=5)

print(f"Cross-validation scores: {cv_scores}")
print(f"Mean CV accuracy: {np.mean(cv_scores):.2f}")
```

Bagging classifier

```python
# from sklearn.ensemble import BaggingClassifier

# Create a bagging classifier with the decision tree pipeline
bagging_classifier = BaggingClassifier(pipeline, n_estimators=10, random_state=42)

# Train the bagging classifier on the training data
bagging_classifier.fit(X_train3, y_train3)
```

```python
# Make prediction on the testing data
y_pred_bagging = bagging_classifier.predict(X_test3)

# Classification Report
print(classification_report(y_pred_bagging, y_test3))
```

```python
accuracy_bagging = accuracy_score(y_test3, y_pred_bagging)
precision_bagging = precision_score(y_test3, y_pred_bagging)
recall_bagging = recall_score(y_test3, y_pred_bagging)

print("Accuracy:", accuracy_bagging)
print("Precision:", precision_bagging)
print("Recall:", recall_bagging)
```

```python
# Evaluate the classifier using cross-validation
# X_train3 should be X and y_train3 should be y.
cv_scores = cross_val_score(bagging_classifier, X_train3, y_train3, cv=5)

print(f"Cross-validation scores: {cv_scores}")
print(f"Mean CV accuracy: {np.mean(cv_scores):.2f}")
```
